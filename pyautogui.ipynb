```python
import pyautogui

# Enable the corner fail-safe (enabled by default)
pyautogui.FAILSAFE = True

# Add a mandatory 0.5-second pause after every PyAutoGUI call to slow it down
pyautogui.PAUSE = 0.5
```

```python
# Get screen resolution (e.g., Returns 1920, 1080)
width, height = pyautogui.size()
```

```python
print(f"Screen resolution: {width}x{height}")
```

```python

# Get current mouse coordinates
x, y = pyautogui.position()
print(f"Current mouse position: ({x}, {y})")
```

```python
# Move the mouse to a specific position (e.g., 100, 100)
print("Moving mouse to (x, y)...")
pyautogui.moveTo(x, y, duration=0.5)  # Move to (x, y) over 1.25 seconds

# Clicking on current position
print("Clicking on current mouse position...")
pyautogui.click(x, y)
```

```python

# Move the mouse to coordinates (100, 150) over 2 seconds
pyautogui.moveTo(100, 150, duration=2)
```

```python

# Move the mouse relative to its current position (right 50 pixels, down 0)
pyautogui.move(50, 0)

# Left-click at the current position
pyautogui.click()

# Right-click at specific coordinates
pyautogui.click(x=200, y=300, button='right')

# Double-click the mouse
pyautogui.doubleClick()

# Drag the mouse to a new position while holding the click
pyautogui.dragTo(500, 500, duration=1)
```
